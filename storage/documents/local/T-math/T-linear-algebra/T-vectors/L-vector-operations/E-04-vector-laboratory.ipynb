# Visual laboratory: vector operations

The same coordinate operations can be read as geometric constructions. We inspect addition, relative displacement, scalar multiplication, and projection in $\mathbb{R}^2$.

## Addition composes displacements

For $\vec u=(3,1)$ and $\vec v=(1,2)$, the diagonal of the parallelogram is $\vec u+\vec v=(4,3)$.

```python
import numpy as np
import matplotlib.pyplot as plt

paper = '#fbf4e8'
ink = '#30241d'
purple = '#6c4ccf'
coral = '#dd6b55'
teal = '#3d8c7a'
grid = '#d9cfc1'

def arrow(ax, origin, vec, color, label, width=0.045):
    ax.arrow(*origin, *vec, length_includes_head=True, head_width=.18, head_length=.22, linewidth=2, width=width, color=color)
    end = np.asarray(origin) + np.asarray(vec)
    ax.text(end[0]+.12, end[1]+.12, label, color=color, fontsize=13, weight='bold')

u = np.array([3., 1.])
v = np.array([1., 2.])
fig, ax = plt.subplots(figsize=(7, 6), facecolor=paper)
ax.set_facecolor(paper)
arrow(ax, (0,0), u, purple, r'$\vec u$')
arrow(ax, (0,0), v, coral, r'$\vec v$')
arrow(ax, (0,0), u+v, teal, r'$\vec u+\vec v$')
ax.plot([u[0], u[0]+v[0]], [u[1], u[1]+v[1]], '--', color=coral, alpha=.75)
ax.plot([v[0], v[0]+u[0]], [v[1], v[1]+u[1]], '--', color=purple, alpha=.75)
ax.axhline(0, color=ink, linewidth=1); ax.axvline(0, color=ink, linewidth=1)
ax.set_xlim(-.5, 4.8); ax.set_ylim(-.5, 4.1); ax.set_aspect('equal')
ax.grid(color=grid, alpha=.6); ax.set_title('The parallelogram law', color=ink, fontsize=17, weight='bold')
for s in ax.spines.values(): s.set_visible(False)
plt.show()
```

## Scaling controls magnitude and orientation

All vectors $\alpha\vec v$ remain on $\operatorname{span}\{\vec v\}$. The absolute value $|\alpha|$ controls magnitude; its sign controls orientation.

```python
base = np.array([2., 1.])
scales = [-1.2, .5, 1., 1.7]
colors = [coral, '#dfaa3f', purple, teal]
fig, ax = plt.subplots(figsize=(9, 5), facecolor=paper)
ax.set_facecolor(paper)
for alpha, color in zip(scales, colors):
    arrow(ax, (0,0), alpha*base, color, rf'${alpha:g}\vec v$', width=.025)
line = np.linspace(-3, 4, 2)
ax.plot(2*line, line, color=ink, linewidth=1.2, alpha=.35)
ax.axhline(0, color=ink, linewidth=1); ax.axvline(0, color=ink, linewidth=1)
ax.set_xlim(-3.2, 4.2); ax.set_ylim(-2, 2.5); ax.set_aspect('equal')
ax.grid(color=grid, alpha=.55); ax.set_title('One span, many scalar weights', color=ink, fontsize=17, weight='bold')
for s in ax.spines.values(): s.set_visible(False)
plt.show()
```

## Projection extracts the component along a direction

The vector projection of $\vec v$ onto $\vec u$ is

$$
\operatorname{proj}_{\vec u}(\vec v)=\frac{\vec v\cdot\vec u}{\vec u\cdot\vec u}\vec u.
$$

```python
u = np.array([3., 1.])
v = np.array([1., 3.])
proj = (v @ u) / (u @ u) * u
fig, ax = plt.subplots(figsize=(7, 6), facecolor=paper)
ax.set_facecolor(paper)
arrow(ax, (0,0), u, purple, r'$\vec u$')
arrow(ax, (0,0), v, coral, r'$\vec v$')
arrow(ax, (0,0), proj, teal, r'$\operatorname{proj}_{\vec u}(\vec v)$', width=.03)
ax.plot([proj[0], v[0]], [proj[1], v[1]], '--', color=ink, alpha=.65)
ax.scatter([proj[0]], [proj[1]], color=teal, s=40)
ax.axhline(0, color=ink, linewidth=1); ax.axvline(0, color=ink, linewidth=1)
ax.set_xlim(-.4, 3.8); ax.set_ylim(-.4, 3.8); ax.set_aspect('equal')
ax.grid(color=grid, alpha=.55); ax.set_title('Projection as directional content', color=ink, fontsize=17, weight='bold')
for s in ax.spines.values(): s.set_visible(False)
plt.show()
```
