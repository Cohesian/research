# Visual laboratory: division as geometry

This laboratory gives the same quotient several visible meanings. We use

$$
\frac{1}{5}=0.2
$$

as the common anchor. The first view partitions a whole, the second scales a vector, and the third compares one magnitude with a larger measuring unit.

## One centimetre, five equal pieces

Changing units exposes discrete cut positions:

$$
1\,\mathrm{cm}=10\,\mathrm{mm},
\qquad
\frac{10\,\mathrm{mm}}{5}=2\,\mathrm{mm}.
$$

Five pieces have four interior cuts. Every coloured interval below is one copy of the quotient.

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

ink = '#30241d'
paper = '#fbf4e8'
palette = ['#6c4ccf', '#dd6b55', '#dfaa3f', '#3d8c7a', '#4f77b8']

fig, ax = plt.subplots(figsize=(11, 3.3), facecolor=paper)
ax.set_facecolor(paper)
for k, color in enumerate(palette):
    ax.add_patch(Rectangle((2*k, 0.25), 2, 0.75, facecolor=color, edgecolor=paper, linewidth=3))
    ax.text(2*k + 1, 0.625, f'part {k+1}', ha='center', va='center', color='white', weight='bold')
for x in range(0, 11, 2):
    ax.plot([x, x], [0.12, 1.13], color=ink, linewidth=1.4)
    ax.text(x, 0.02, f'{x} mm', ha='center', va='top', color=ink, fontsize=10)
ax.annotate('', xy=(10, 1.5), xytext=(0, 1.5), arrowprops=dict(arrowstyle='<->', color=ink, linewidth=2))
ax.text(5, 1.62, 'the same whole: 10 mm = 1 cm', ha='center', color=ink, fontsize=13, weight='bold')
ax.text(5, -0.42, 'each part: 2 mm = 0.2 cm', ha='center', color=ink, fontsize=13)
ax.set_xlim(-0.5, 10.5)
ax.set_ylim(-0.65, 1.95)
ax.axis('off')
plt.show()
```

## Division by five as a linear scaling

For a directed segment $\vec a$, division by five is multiplication by $1/5$:

$$
S_{1/5}(\vec a)=\frac{1}{5}\vec a.
$$

The operation preserves direction and reduces magnitude. Applying the inverse scale reconstructs the original vector.

```python
fig, ax = plt.subplots(figsize=(10, 3.6), facecolor=paper)
ax.set_facecolor(paper)
ax.annotate('', xy=(10, 2.3), xytext=(0, 2.3), arrowprops=dict(arrowstyle='-|>', color=palette[0], linewidth=6))
ax.text(5, 2.65, r'$\vec{a}$   magnitude $1\,\mathrm{cm}$', ha='center', color=ink, fontsize=14)
ax.annotate('', xy=(2, 0.75), xytext=(0, 0.75), arrowprops=dict(arrowstyle='-|>', color=palette[1], linewidth=6))
ax.text(1, 1.1, r'$\frac{1}{5}\vec{a}$   magnitude $0.2\,\mathrm{cm}$', ha='center', color=ink, fontsize=14)
ax.annotate(r'$S_{1/5}$', xy=(4.8, 1.0), xytext=(4.8, 2.05), ha='center', color=ink, fontsize=14, arrowprops=dict(arrowstyle='->', color=ink, connectionstyle='arc3,rad=.25'))
ax.annotate(r'$S_{5}$ reconstructs', xy=(6.0, 2.15), xytext=(6.0, 1.0), ha='center', color=ink, fontsize=13, arrowprops=dict(arrowstyle='->', color=ink, connectionstyle='arc3,rad=.25'))
ax.set_xlim(-0.5, 10.8)
ax.set_ylim(0, 3.2)
ax.axis('off')
plt.show()
```

## One magnitude measured by another

In the quotative reading, both operands carry the same unit:

$$
\frac{1\,\mathrm{cm}}{5\,\mathrm{cm}}=0.2.
$$

The quotient says that the shorter segment occupies twenty percent of the reference segment. Formally, the reference-normalization map $N_b(x)=x/b$ sends $5\,\mathrm{cm}$ to $1$ and $1\,\mathrm{cm}$ to $0.2$. On a percentage scale, the same pair becomes $100$ and $20$.

```python
fig, ax = plt.subplots(figsize=(10, 3.5), facecolor=paper)
ax.set_facecolor(paper)
ax.add_patch(Rectangle((0, 1.8), 5, 0.65, facecolor='#d9cfc1', edgecolor='none'))
ax.add_patch(Rectangle((0, 1.8), 1, 0.65, facecolor=palette[0], edgecolor='none'))
ax.text(2.5, 2.75, r'reference unit $b=5\,\mathrm{cm}$', ha='center', color=ink, fontsize=14, weight='bold')
ax.text(0.5, 2.125, r'$a=1\,\mathrm{cm}$', ha='center', va='center', color='white', fontsize=12, weight='bold')
ax.text(2.5, 1.3, r'$a/b=0.2=20\%$', ha='center', color=ink, fontsize=17, weight='bold')
for x in range(6):
    ax.plot([x, x], [1.72, 2.53], color=paper, linewidth=2)
ax.set_xlim(-0.35, 5.35)
ax.set_ylim(0.7, 3.2)
ax.axis('off')
plt.show()
```

## What remains invariant?

Each view preserves the reconstruction equation:

$$
1=5(0.2).
$$

What changes is the semantic role of the denominator: a count of parts, a reciprocal scale, or a measuring unit. Division becomes easier to reason about when the equation, the units, and the question are read together.
